# Question 5: Calculate the sum of digits in number 12345


```python
number=12345
reversed=0
digit=0
while number>0:
    digit=number%10
    reversed += digit
    number //= 10
print(reversed)      
```

# Question 9: Find the greatest common divisor (GCD) of 48 and 36


```python
n=48
m=36
for i in range(1,min(n,m)+1):
    if m % i== 0 and n % i ==0:
        gcd=i
print(gcd)        
```
