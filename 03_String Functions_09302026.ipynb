# Question 41: Check if string contains balanced parentheses: "((()))"


```python
String="((()))"
if String.count("(") == String.count(")"):
    print("String contains balanced parentheses")
else:
    print("String not balance")    
```

```python
String="((()))"
count=0
count1=0
for i in String:
    if i =="(":
        count += 1
    else:
        count1 += 1
    if count <0:
        break
if count == count1:
      print("String contains balanced parentheses")
else:
    print("String not balanced")
```

# Question 50: Check if string is a valid credit card number: "4532015112830366"


```python
String="4532015112830366"
if String.isdigit() and len(String) == 16 :
    print("The card number is valid")
else :
    print("The Card number is not valid")    
```

```python
String = "4532015112830366"

if String.isdigit() and len(String) == 16:
    total_sum = 0
    # Reverse the string to process from right to left
    reversed_digits = String[::-1]

    for index, digit in enumerate(reversed_digits):
        num = int(digit)

        # Double every second digit (odd indices in the 0-indexed reversed list)
        if index % 2 == 1:
            num *= 2
            # If doubling results in a 2-digit number, subtract 9 (same as adding digits)
            if num > 9:
                num -= 9

        total_sum += num

    # If the total sum is perfectly divisible by 10, the card is valid
    if total_sum % 10 == 0:
        print("The card number is valid")
    else:
        print("The Card number is not valid")
else:
    print("The Card number is not valid")

# Output: The Card number is not valid
```
