# Question 1: Calculate the average of 3.14, 2.718, 1.618, 0.577


```python
numbers=3.14,2.718,1.618,0.577
average=sum(numbers)/len(numbers)
print(average)
```

# Question 3: Calculate the compound interest on $1000 at 5.5% for 3 years


```python
P=1000
R=5.5
T=3
A=P*((1+(R/100))**T)
CI=A-P
print(A,CI)
```
