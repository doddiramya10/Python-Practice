4. Hollow square


```python
n = 5
for i in range(n):
    if i == 0 or i == n - 1:
        print("*" * n)
    else:
        print("*" + " " * (n-2) + "*")
```

6. Floyd’s triangle (continuous numbers)


```python
n=4
num=1
for i in range(1,n+1):
  for j in range(i):
    print(num,end="")
    num+=1
  print()
```

8. Inverted centered star pyramid


```python
n=5
for i in range(5,0,-1):
    spaces = " " * (n-i)
    l_stars = "*"*i
    r_stars = "*" * (i-1)
    print(spaces+l_stars+r_stars)
```

9. Diamond (centered, stars)


```python
n= 5
# upper triangle
for i in range(1,n+1):
    spaces = " " * (n-i)
    l_stars = "*"* i
    r_stars="*"*(i-1)
    print(spaces+r_stars+l_stars)

# lower Triangle
for i in range(5,0,-1):
    d_space = " " * (n-i)
    dl_stars = "*"* i
    dr_stars = "*" * (i-1)
    print(d_space+dl_stars+dr_stars)
```

11. Binary triangle (0 and 1)


```python
n=5
for i in range(1,n+1):
    spaces = " "* i
    stars = str(1) * i
    print(stars+spaces)
```

```python
n=5
for i in range(1,n+1):
    spaces = " " * (n-i)
    if i == 1:
      stars = "1"
      r_stars = ""
    else:
      stars ="1"
      r_stars = "01" *(i-1)
    print(spaces + stars + r_stars)

```

```python
n= 5
for i in range(1,n+1):
    # spaces = " " * (n-i)
    if i ==1 :
        stars = "1"* i
    else:
        stars = "01"*(i //2) if i% 2 == 0 else "10"* (i//2)+"1"
    print(stars)
```

12. Checkerboard


```python
n = 5
for i in range(1,n+1):
  if i ==1 :
    hash = "#"*1
    stars = 1*"."
    print(hash+stars+hash+stars)
  else:
    stars = ".#"* (i//2) if i % 2 ==0 else "#." *(i//2)
    print(stars)
```
