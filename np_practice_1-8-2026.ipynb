```python
arr = np.array([5, 12, 18, 25, 30, 35, 42, 50, 55, 60, 68, 75, 80, 95, 100])
```

```python
numbers=arr[(arr<=20) | (arr>=80)]
print(numbers)
```

```python
#Replace all numbers divisible by 10 with 999.
numbers=arr[arr %10]=999
print(arr)
```

```python
#Print numbers that are divisible by 2 and 3.
numbers=arr[(arr%2==0) & (arr%3==0)]
print(numbers)
```

```python
#Print numbers that are not divisible by 5.
numbers = arr[arr % 5 !=0]
print(numbers)
```

```python
#Replace all numbers between 40 and 80 with 100.
numbers=arr[(arr>=40) & (arr<=80)]=100
print(arr)
```

```python
#Print numbers that are greater than 30 and are even.
numbers=arr[(arr>30) & (arr %2 ==0)]
print(numbers)
```
